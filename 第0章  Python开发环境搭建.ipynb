安装完Python后，我们来启动Python。这里介绍两种工具的启动方法，分别是IDLE的启动方法和VS Code的启动方法。
先介绍IDLE的启动方法。提前说明一下，这里用的是Windows，不过没关系，因为Mac OS和Linux的启动方法与Windows的方法类似。在开始菜单中，找到Python的IDLE选项后打开，就进入了Python。可以参见图1。

另一种方法是在VS Code左侧的边栏中点击扩展，搜索Python插件，安装完毕后即可。这样可以让VS Code支持运行Python。另外，我们还可以用同样的方式安装Pylance插件Jupyter插件。Jupyter插件可以让我们方便学习和实验,Pylance插件可以给我们提供代码提示。可以参见图2。


当然，这也只能运行简单的代码。如果要编写代码比较长，这样就不合适了。我们需要进入Python编辑器去编写代码。
在已经打开的Python shell（IDLE）里点击上方的Files,选择New File，就会有一个新窗口叫untitled，在这里编写完代码保存后就可以运行了。
在VS Code中，用同样的方式点击上方的Files后点击New File新建，保存后会在VS Code的终端上运行。
打开完毕后，我们可以尝试运行一下第一条代码：

```python
print("Hello Pyhton!")
```
